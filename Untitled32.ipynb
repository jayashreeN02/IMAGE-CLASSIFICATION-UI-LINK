```python

import gradio as gr
import torch
from PIL import Image, ImageOps
from transformers import BlipProcessor, BlipForConditionalGeneration
import html

# ==========================================
# 1. MODEL INITIALIZATION
# ==========================================

device = "cuda" if torch.cuda.is_available() else "cpu"

if "processor" not in globals() or "model" not in globals():
    print("Loading VisionAI model...")

    processor = BlipProcessor.from_pretrained(
        "Salesforce/blip-image-captioning-base"
    )

    model = BlipForConditionalGeneration.from_pretrained(
        "Salesforce/blip-image-captioning-base"
    ).to(device)

    model.eval()

print("VisionAI is ready!")


# ==========================================
# 2. IMAGE ANALYSIS
# ==========================================

def analyze_image(image, category):

    if image is None:
        return None, "<h3>⚠ Please upload an image first.</h3>", "Waiting for image"

    try:
        image = ImageOps.exif_transpose(image).convert("RGB")

        inputs = processor(
            images=image,
            return_tensors="pt"
        ).to(device)

        with torch.inference_mode():
            output = model.generate(
                **inputs,
                max_new_tokens=55
            )

        description = processor.decode(
            output[0],
            skip_special_tokens=True
        )

        safe_description = html.escape(description.capitalize())
        safe_category = html.escape(category)

        result = f"""
        <div class="result-card">
            <div class="success-label">
                <span class="pulse"></span>
                ANALYSIS COMPLETE
            </div>

            <h2>✨ Here's what I discovered</h2>

            <div class="description-box">
                <p>{safe_description}</p>
            </div>

            <div class="metadata">
                <div class="meta-item">
                    <span>Selected category</span>
                    <strong>{safe_category}</strong>
                </div>

                <div class="meta-item">
                    <span>AI Engine</span>
                    <strong>BLIP Vision</strong>
                </div>

                <div class="meta-item">
                    <span>Processing device</span>
                    <strong>{device.upper()}</strong>
                </div>
            </div>

            <div class="disclaimer">
                AI-generated description. The selected category
                is a preference, not a verified classification.
            </div>
        </div>
        """

        return image, result, "✓ Analysis completed successfully"

    except Exception as e:
        return (
            image,
            f"<h3>Analysis failed</h3><p>{html.escape(str(e))}</p>",
            "Something went wrong"
        )


# ==========================================
# 3. PREMIUM CSS
# ==========================================

css = """
@import url('https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700;800&display=swap');

.gradio-container {
    max-width: 1250px !important;
    margin: auto !important;
    font-family: 'Inter', sans-serif !important;
    background: #090d1c !important;
    color: #f4f4ff !important;
}

body {
    background: #090d1c !important;
}

footer {
    display: none !important;
}

#hero {
    background:
        radial-gradient(circle at 85% 20%, #7045b8 0%, transparent 35%),
        linear-gradient(120deg, #151b36, #241d49 65%, #382765);
    border: 1px solid #51417a;
    border-radius: 28px;
    padding: 48px;
    margin: 20px 0 28px;
    box-shadow: 0 20px 60px #0005;
}

#brand {
    color: #c8b5ff;
    font-size: 12px;
    font-weight: 800;
    letter-spacing: 3px;
    margin-bottom: 22px;
}

#hero h1 {
    color: white;
    font-size: clamp(38px, 6vw, 65px);
    font-weight: 800;
    line-height: 1.08;
    letter-spacing: -2px;
    margin-bottom: 18px;
}

#hero p {
    color: #c5c5e3;
    font-size: 15px;
    line-height: 1.9;
}

.hero-tags {
    display: flex;
    flex-wrap: wrap;
    gap: 10px;
    margin-top: 24px;
}

.hero-tags span {
    background: #ffffff12;
    border: 1px solid #ffffff25;
    border-radius: 30px;
    padding: 9px 15px;
    color: #eeeeff;
    font-size: 12px;
}

.panel {
    background: #151a2d !important;
    border: 1px solid #292e48 !important;
    border-radius: 22px !important;
    padding: 25px !important;
    box-shadow: 0 12px 35px #0003;
}

.panel h2, .panel h3 {
    color: #f3f0ff !important;
}

.panel p, .panel label {
    color: #c3c5dc !important;
}

.upload-panel {
    border: 1px dashed #7963cc !important;
    border-radius: 16px !important;
    background: #1c2039 !important;
}

#analyze {
    border: none !important;
    border-radius: 13px !important;
    height: 55px;
    font-weight: 800 !important;
    font-size: 15px !important;
    background: linear-gradient(100deg, #7355ed, #a16af7) !important;
    color: white !important;
    box-shadow: 0 8px 25px #7955e955;
    transition: .25s ease;
}

#analyze:hover {
    transform: translateY(-2px);
    filter: brightness(1.15);
}

#clear {
    border-radius: 12px !important;
}

#result {
    background: #151a2d !important;
    border: 1px solid #292e48 !important;
    border-radius: 18px !important;
    padding: 20px !important;
    min-height: 280px;
    color: white !important;
}

.result-card h2 {
    color: #f5f1ff;
    font-size: 25px;
    margin: 18px 0;
}

.success-label {
    color: #79e7b1;
    font-size: 11px;
    font-weight: 800;
    letter-spacing: 2px;
}

.pulse {
    display: inline-block;
    width: 8px;
    height: 8px;
    background: #5de6a4;
    border-radius: 50%;
    margin-right: 7px;
    box-shadow: 0 0 10px #5de6a4;
}

.description-box {
    background: linear-gradient(120deg, #292448, #25223c);
    border: 1px solid #48406c;
    border-radius: 15px;
    padding: 20px;
    margin: 18px 0;
}

.description-box p {
    color: #e9e4ff !important;
    font-size: 16px;
    line-height: 1.9;
}

.metadata {
    margin-top: 20px;
}

.meta-item {
    display: flex;
    justify-content: space-between;
    gap: 12px;
    padding: 14px 0;
    border-bottom: 1px solid #30334a;
    font-size: 13px;
}

.meta-item span {
    color: #999fbe;
}

.meta-item strong {
    color: #d7c7ff;
    text-align: right;
}

.disclaimer {
    margin-top: 22px;
    color: #858ba8;
    font-size: 11px;
    line-height: 1.8;
}

#status textarea {
    text-align: center;
    color: #80e5b5 !important;
    font-weight: 700;
    background: #1b2630 !important;
    border-color: #344e49 !important;
}

footer-area {
    text-align: center;
}

.section-heading {
    font-size: 19px;
    font-weight: 800;
    color: #eeeaff;
    margin-bottom: 20px;
}

#footer {
    text-align: center;
    padding: 30px 10px;
    color: #858ba8;
    font-size: 12px;
    line-height: 2;
}

@media(max-width: 700px) {
    #hero {
        padding: 27px;
    }

    .panel {
        padding: 15px !important;
    }

    #hero h1 {
        font-size: 42px;
    }
}
"""


# ==========================================
# 4. BUILD VISIONAI WEBSITE
# ==========================================

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="violet",
        secondary_hue="indigo",
        neutral_hue="slate"
    ),
    css=css,
    title="VisionAI | AI Image Recognition"
) as demo:

    # HERO SECTION

    gr.HTML("""
    <div id="hero">

        <div id="brand">✦ VISIONAI STUDIO</div>

        <h1>
            Discover the<br>
            world through AI.
        </h1>

        <p>
            Your intelligent visual companion.
            Upload any image and let artificial intelligence
            transform pixels into meaningful descriptions.
        </p>

        <div class="hero-tags">
            <span>🐾 Animals</span>
            <span>🍎 Fruits</span>
            <span>💻 Electronics</span>
            <span>🔤 Alphabet</span>
            <span>🌿 Nature</span>
        </div>

    </div>
    """)

    with gr.Row(equal_height=False):

        # LEFT PANEL

        with gr.Column(
            scale=5,
            elem_classes=["panel"]
        ):

            gr.Markdown("## 📤 Upload your image")

            image_input = gr.Image(
                type="pil",
                label="Drop your image here or browse",
                height=300,
                elem_classes=["upload-panel"]
            )

            category_input = gr.Dropdown(
                choices=[
                    "All Categories",
                    "Animals",
                    "Fruits",
                    "Electronics",
                    "Alphabet",
                    "Vehicles",
                    "Flowers and Nature",
                    "Other Objects"
                ],
                value="All Categories",
                label="🎯 Select image category"
            )

            analyze_button = gr.Button(
                "✦  Analyze Image with AI",
                variant="primary",
                elem_id="analyze"
            )

            clear_button = gr.ClearButton(
                components=[image_input],
                value="↻  Clear Image",
                elem_id="clear"
            )

            gr.Markdown("""
            ### 💡 Tips for better results

            - Upload a clear, well-lit image.
            - Keep the main object visible.
            - Avoid blurry photographs.
            - Use large readable letters for alphabet images.
            """)

        # RIGHT PANEL

        with gr.Column(
            scale=6,
            elem_classes=["panel"]
        ):

            gr.Markdown("## ✨ AI Recognition Results")

            image_preview = gr.Image(
                label="Image Preview",
                height=210,
                interactive=False
            )

            result_output = gr.HTML(
                value="""
                <div class="result-card">
                    <div class="success-label">
                        ✦ YOUR AI WORKSPACE
                    </div>

                    <h2>Ready to explore?</h2>

                    <div class="description-box">
                        <p>
                            Your image analysis results will
                            appear here. Upload an image and
                            click Analyze Image with AI to begin.
                        </p>
                    </div>
                </div>
                """,
                elem_id="result"
            )

            status_output = gr.Textbox(
                value="● System ready",
                label="System Status",
                interactive=False,
                elem_id="status"
            )

    # FOOTER

    gr.HTML("""
    <div id="footer">
        <strong style="color:#b9a3ff;letter-spacing:2px;">
            ✦ VISIONAI STUDIO
        </strong>
        <br>
        Powered by Python, Gradio and BLIP Vision AI
        <br>
        AI-generated descriptions may occasionally be inaccurate.
        <br>
        © 2026 VisionAI Studio
    </div>
    """)

    # BUTTON EVENT

    analyze_button.click(
        fn=analyze_image,
        inputs=[image_input, category_input],
        outputs=[
            image_preview,
            result_output,
            status_output
        ]
    )


# ==========================================
# 5. LAUNCH APPLICATION
# ==========================================

demo.queue()

demo.launch(
    share=True,
    show_error=True
)
```
